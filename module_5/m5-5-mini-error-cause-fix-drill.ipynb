## Why Drill Error → Cause → Fix?

In the previous mini-labs and the wall lab, you walked through failures step by step — with the code in front of you, the traceback visible, and time to investigate.

In real work, you don't have that luxury. A `KeyError` flashes on screen and you need to **instantly** know: *"Wrong key name or missing key — check the actual keys with `.keys()` or `print(data)`."* That instant recognition comes from **retrieval practice** — seeing an error and recalling the cause and fix from memory.

This mini-lab is a **drill**. You'll see six error scenarios — a mix of loud and silent failures. For each one, you'll fill in the cause and fix **before** revealing the answer. Then we'll compile everything into a debug log.

````python
from IPython.display import display, Markdown

def md(text):
    """Display text as rendered markdown."""
    display(Markdown(text))

# Our debug log — we'll build it drill by drill
debug_log = []
````

## How This Drill Works

For each scenario:

1. **Read** the error message or observe the wrong output
2. **Pause** — try to recall the cause and fix from memory before running the next cell
3. **Run** the answer cell to check yourself
4. **Log** the entry into the debug log

The goal isn't to debug live code — you already did that. The goal is **fast recall**: error → cause → fix, without Googling.

```
ModuleNotFoundError: No module named 'pandas'
```

**Before running the next cell**, recall:
- What is the most likely **cause**?
- What is the **fix**?
- Is this a **loud** or **silent** failure?

````python
# Drill 1 — Answer
drill_1 = {
    "error": "ModuleNotFoundError: No module named 'pandas'",
    "type": "Loud (traceback)",
    "cause": "Package not installed in the active Python environment",
    "fix": "Run: python -m pip install pandas (use sys.executable to target the right interpreter)",
}

md(f"""**Type:** {drill_1['type']}  
**Cause:** {drill_1['cause']}  
**Fix:** `{drill_1['fix']}`

**Key insight:** Always install with the *same* Python interpreter the notebook uses — `sys.executable -m pip install`.""")

debug_log.append(drill_1)
````

## Drill 3 — Silent: Empty Filter Result

Your code runs without error, but:

```python
prices = [{"item": "A", "price": "50"}, {"item": "B", "price": "120"}]
expensive = [p for p in prices if p["price"] > 100]
print(len(expensive))  # prints: 0
```

You expected 1 result (item B). You got 0. No traceback.

**Recall:** cause, fix, loud or silent?

````python
# Drill 3 — Answer
drill_3 = {
    "error": "Silent: filter returns 0 results instead of 1, no traceback",
    "type": "Silent (no traceback)",
    "cause": "Price is a string from CSV/JSON source; string > int comparison fails or gives wrong result",
    "fix": "Convert to numeric before comparing: int(p['price']) > 100",
}

md(f"""**Type:** {drill_3['type']}  
**Cause:** {drill_3['cause']}  
**Fix:** `{drill_3['fix']}`

**Key insight:** Always check `type(value)` and `repr(value)` when a filter returns unexpected results.""")

debug_log.append(drill_3)
````

You see this traceback:

```
NameError: name 'reuslt' is not defined
```

**Recall:** cause, fix, loud or silent?

````python
# Drill 4 — Answer
drill_4 = {
    "error": "NameError: name 'reuslt' is not defined",
    "type": "Loud (traceback)",
    "cause": "Typo in variable name — 'reuslt' instead of 'result'",
    "fix": "Fix the typo: change 'reuslt' to 'result'",
}

md(f"""**Type:** {drill_4['type']}  
**Cause:** {drill_4['cause']}  
**Fix:** `{drill_4['fix']}`

**Key insight:** `NameError` almost always means a typo or using a variable before it's assigned. Read the name character by character.""")

debug_log.append(drill_4)
````

## Drill 5 — Silent: Wrong Aggregation

Your code runs without error, but:

```python
scores = [85, 92, 78, 90]
average = sum(scores) / len(scores) + 1   # meant to add 1 bonus point THEN average
print(average)  # prints: 87.25
```

You expected the average of `[86, 93, 79, 91]` = **87.25**. The output happens to be 87.25 too — but the logic is wrong. With different data, the bug surfaces.

Actually, let's check: `sum([85,92,78,90]) / 4 + 1 = 345/4 + 1 = 86.25 + 1 = 87.25`. The correct calculation `sum([86,93,79,91]) / 4 = 349/4 = 87.25`. Same number, different logic!

**Recall:** What category of bug is this? Why is it dangerous?

````python
# Drill 5 — Answer
drill_5 = {
    "error": "Silent: operator precedence gives right answer for this data but wrong logic",
    "type": "Silent (no traceback, coincidentally correct output)",
    "cause": "Operator precedence: division happens before addition, so +1 is added to the average, not to each score before averaging",
    "fix": "Use parentheses to make intent explicit: sum(s + 1 for s in scores) / len(scores)",
}

md(f"""**Type:** {drill_5['type']}  
**Cause:** {drill_5['cause']}  
**Fix:** `{drill_5['fix']}`

**Key insight:** The scariest silent bugs are ones where the wrong logic produces the right answer *for your test data*. \
Always test with data where the wrong approach gives a *different* number than the right approach.""")

debug_log.append(drill_5)
````

## Drill 6 — `AttributeError` (Name Shadowing)

You see this traceback:

```
AttributeError: module 'email' has no attribute 'mime'
```

You have a file called `email.py` in your project folder.

**Recall:** cause, fix, loud or silent?

````python
# Drill 6 — Answer
drill_6 = {
    "error": "Silent: module 'email' loads your local file instead of stdlib, then AttributeError on submodule access",
    "type": "Silent (misleading traceback — the real cause has no error)",
    "cause": "A local file named email.py shadows the standard library email package",
    "fix": "Rename or delete the local email.py, clear __pycache__, and re-import",
}

md(f"""**Type:** {drill_6['type']}  
**Cause:** {drill_6['cause']}  
**Fix:** `{drill_6['fix']}`

**Key insight:** Check `module.__file__` whenever a standard library module behaves strangely. \
If it points to your project folder, you have a shadowing problem.""")

debug_log.append(drill_6)
````

## Review: The Complete Debug Log

Here's every error → cause → fix from this drill, compiled into a single table. This is the kind of personal reference that, over time, eliminates re-Googling the same failures.

````python
rows = "| # | Type | Error | Cause | Fix |\n"
rows += "|---|------|-------|-------|-----|\n"
for i, entry in enumerate(debug_log, 1):
    rows += f"| {i} | {entry['type'][:12]} | {entry['error'][:45]} | {entry['cause'][:50]} | {entry['fix'][:50]} |\n"

md(rows)
````

## The Debug Log as a Habit

The drills above are synthetic — but the **habit** is real. Every time you hit a failure in your own work:

1. **Record** the error message (or describe the silent symptom)
2. **Record** the root cause you discovered
3. **Record** the one change that fixed it

After a few weeks, your log becomes a personal cheat sheet. After a few months, you stop needing it — the pattern recognition becomes automatic.

| When | What to Log |
|------|------------|
| Loud failure | Copy the last line of the traceback as the error |
| Silent failure | Describe the symptom: "filter returned 0 rows" or "average was wrong" |
| Cause | One sentence: what was actually wrong |
| Fix | One sentence: the single change you made |

## 🎯 Summary

### Key Takeaways

1. **Fast recall beats slow investigation** — drilling error → cause → fix from memory builds the pattern recognition that makes debugging fast in real work
2. **Half of common bugs are silent** — they produce no traceback, just wrong output; you catch them only by knowing what the correct output should be
3. **A debug log compounds over time** — recording every failure you fix turns painful debugging sessions into a reusable personal reference
4. **Three fields are enough** — error (or symptom), cause, fix; keep entries short so the log stays useful and searchable
