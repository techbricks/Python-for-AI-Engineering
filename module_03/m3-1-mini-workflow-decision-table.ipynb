| Workflow Form Choice | Explore in a notebook, automate once in a script, reuse reliably as a package with tests, or expose it as a service that other systems can call. |


## Why This Matters

AI engineers constantly switch between exploration and production. Picking the **wrong form** for the job wastes time:

- Exploring data in a full Python package → too much boilerplate
- Deploying a model from a Jupyter notebook → fragile, hard to schedule

A simple decision table saves you from these traps.

## The Four Workflow Forms

| Form | What It Is | Best For |
|------|-----------|----------|
| **Notebook** | `.ipynb` — cells of code + markdown | Exploration, prototyping, teaching, EDA |
| **Script** | `.py` file run from the terminal | Automation, pipelines, cron jobs, CLI tools |
| **Package** | Folder with `__init__.py`, installable | Reusable library code shared across projects |
| **Service** | Long-running process (API, worker) | Serving predictions, handling requests 24/7 |

````python
from IPython.display import display, Markdown

def md(text):
    """Display text as rendered markdown."""
    display(Markdown(text))
````

## Building the Decision Table

We'll encode the decision table as a simple data structure so we can query it programmatically.

````python
# Decision table: each form with its traits
DECISION_TABLE = {
    "notebook": {
        "extension": ".ipynb",
        "interactive": True,
        "schedulable": False,
        "importable": False,
        "serves_http": False,
        "best_for": ["exploration", "prototyping", "teaching", "EDA", "visualization"],
    },
    "script": {
        "extension": ".py",
        "interactive": False,
        "schedulable": True,
        "importable": True,
        "serves_http": False,
        "best_for": ["automation", "CLI tools", "data pipelines", "batch jobs"],
    },
    "package": {
        "extension": ".py (folder)",
        "interactive": False,
        "schedulable": False,
        "importable": True,
        "serves_http": False,
        "best_for": ["shared library", "reusable utilities", "team code"],
    },
    "service": {
        "extension": ".py (server)",
        "interactive": False,
        "schedulable": False,
        "importable": True,
        "serves_http": True,
        "best_for": ["API endpoints", "model serving", "webhooks", "real-time inference"],
    },
}

print(f"Forms defined: {list(DECISION_TABLE.keys())}")
````

## Querying the Decision Table

Let's write a small helper that recommends a form based on what you need.

````python
def recommend_form(need: str) -> list[str]:
    """Return form names whose 'best_for' list contains the need (case-insensitive)."""
    need_lower = need.lower()
    matches = []
    for form, traits in DECISION_TABLE.items():
        if any(need_lower in use.lower() for use in traits["best_for"]):
            matches.append(form)
    return matches
````

## Scenario Practice

Let's run through real AI engineering scenarios and see what the table recommends.

````python
scenarios = [
    "I need to explore a new dataset and visualize distributions",
    "I need to run a data cleaning pipeline every night at 2 AM",
    "I need to serve an LLM-powered chatbot API",
    "I need to share embedding utilities across three projects",
    "I need to prototype a prompt and see results immediately",
]

# Map each scenario to a keyword and look it up
scenario_keywords = [
    ("exploration",   "notebook"),
    ("automation",    "script"),
    ("API endpoints", "service"),
    ("shared library","package"),
    ("prototyping",   "notebook"),
]

lines = ["| # | Scenario | Keyword | Recommended Form | ✅ |",
         "|---|----------|---------|-----------------|----|",]

for i, (scenario, (keyword, expected)) in enumerate(zip(scenarios, scenario_keywords), 1):
    result = recommend_form(keyword)
    match = "✅" if expected in result else "❌"
    lines.append(f"| {i} | {scenario} | `{keyword}` | **{result[0]}** | {match} |")

md("\n".join(lines))
````

## Trait Comparison

Let's display all four forms side-by-side so you can see the differences at a glance.

````python
header = "| Trait | Notebook | Script | Package | Service |"
sep    = "|-------|----------|--------|---------|--------|"

traits = ["interactive", "schedulable", "importable", "serves_http"]
labels = ["Interactive?", "Schedulable?", "Importable?", "Serves HTTP?"]

rows = [header, sep]
for label, trait in zip(labels, traits):
    vals = []
    for form in ["notebook", "script", "package", "service"]:
        v = DECISION_TABLE[form][trait]
        vals.append("✅" if v else "—")
    rows.append(f"| {label} | {' | '.join(vals)} |")

md("\n".join(rows))
````

## A Typical AI Project Lifecycle

Most AI engineering projects move through forms over time:

```
notebook → script → package → service
(explore)  (automate) (reuse)   (deploy)
```

Not every project needs all four stages. A one-off analysis might stay as a notebook. A shared utility might stop at package. The key is **choosing the right form for the current stage**.

````python
lifecycle = [
    ("1. Explore & prototype",       "notebook"),
    ("2. Automate the pipeline",     "script"),
    ("3. Extract reusable code",     "package"),
    ("4. Serve to users",            "service"),
]

lines = ["| Stage | Form | Best-For Examples |\n|-------|------|-------------------|"]
for stage, form in lifecycle:
    examples = ", ".join(DECISION_TABLE[form]["best_for"][:3])
    lines.append(f"| {stage} | **{form}** | {examples} |")

md("\n".join(lines))
````
