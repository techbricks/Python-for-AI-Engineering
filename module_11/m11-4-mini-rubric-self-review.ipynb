## Same 9 rows, new evidence column

Score the rows from the start of the module again. Every non-zero score needs an evidence pointer: a path, a metric, or a command output.

```python
from IPython.display import display, Markdown

def md(text):
    """Display text as rendered markdown."""
    display(Markdown(text))
```

```python
BASELINE = [
    {"id": 1, "name": "Pipeline runs end to end", "baseline": 2},
    {"id": 2, "name": "Reusable package", "baseline": 2},
    {"id": 3, "name": "Messy-data ingestion", "baseline": 2},
    {"id": 4, "name": "Reproducible repo", "baseline": 1},
    {"id": 5, "name": "Replaceable AI interface", "baseline": 2},
    {"id": 6, "name": "Hardened runtime", "baseline": 1},
    {"id": 7, "name": "Tests that can fail", "baseline": 1},
    {"id": 8, "name": "Evaluation with numbers", "baseline": 2},
    {"id": 9, "name": "Delivered and honest", "baseline": 0},
]

# After README, LIMITATIONS, and the release audit — still two weak engineering rows
NOW = [
    {"id": 1, "score": 2, "evidence": "src/triage/pipeline.py runs labels on data/sample.csv"},
    {"id": 2, "score": 2, "evidence": "from triage.pipeline import run works from tests/"},
    {"id": 3, "score": 2, "evidence": "prepare.py logs dropped empty=4, too_long=2"},
    {"id": 4, "score": 2, "evidence": "RELEASE_AUDIT.md: documented_commands PASS, pytest PASS"},
    {"id": 5, "score": 2, "evidence": "clients/base.py Protocol; fake adapter in tests"},
    {"id": 6, "score": 1, "evidence": "validate.py exists; bounded retry still missing a wall-clock cap"},
    {"id": 7, "score": 1, "evidence": "test_parser.py passes; no timeout fake-client test yet"},
    {"id": 8, "score": 2, "evidence": "50-row gold set; Billing precision 84% vs baseline 62%"},
    {"id": 9, "score": 2, "evidence": "README four-question order; LIMITATIONS.md demonstration sentence"},
]

# The two lists must describe the same rows in the same order
assert [r["id"] for r in BASELINE] == [r["id"] for r in NOW]

lines = ["| # | Criterion | Baseline | Now | Evidence |", "|---|---|---|---|---|"]
for old, new in zip(BASELINE, NOW):
    lines.append(f"| {old['id']} | {old['name']} | {old['baseline']} | {new['score']} | {new['evidence']} |")

# Lowest score first; ties broken by row number. Used by the next code cell.
rows = [(new["score"], old["id"], old["name"], new["evidence"]) for old, new in zip(BASELINE, NOW)]
weakest_two = sorted(rows)[:2]

md("### Rubric self-review\n\n" + "\n".join(lines))
```

## Fix the two weakest rows

Pick the two lowest rows and make the smallest evidenced fix. Then bump the score.

```python
lines = ["| Weak row | Current score | Concrete fix |", "|---|---|---|"]
for score, rid, name, evidence in weakest_two:
    if rid == 6:
        action = "Add a max_elapsed_s argument to retry() and a test that fails when the cap is removed."
    elif rid == 7:
        action = "Add FakeClient.timeout() and assert the pipeline raises; confirm the test fails if you delete the raise."
    else:
        action = "Write the missing evidence pointer, then the smallest change that makes it true."
    lines.append(f"| {rid} {name} | {score} | {action} |")

md("### Two fixes\n\n" + "\n".join(lines))
```

## Trade-off answers

Answer these four trade-off questions using your repo's numbers. The structure stays.

```python
METRICS = {
    "gold_n": 50,
    "baseline_acc": 0.62,
    "billing_precision": 0.84,
    "p50_ms": 420,
    "p95_ms": 2100,
    "cost_per_1000": 0.42,
}

QUESTIONS = [
    {
        "id": "Q1",
        "ask": "Why is the model behind an interface instead of a vendor SDK call in the pipeline?",
        "rejected": "Interfaces are best practice and more professional.",
        "hired": (
            "Pipeline code types a Protocol, so tests inject FakeClient and cover timeout "
            "and bad JSON without a network or a key."
        ),
    },
    {
        "id": "Q2",
        "ask": "Would you ship this over the baseline?",
        "rejected": "Yes, the model looks more accurate.",
        "hired": (
            f"I would ship this as a demonstration Billing router because Billing precision is "
            f"{METRICS['billing_precision']:.0%} on a {METRICS['gold_n']}-row gold set "
            f"(baseline accuracy {METRICS['baseline_acc']:.0%}), accepting that Account is often "
            f"labelled General and that this is not a production system."
        ),
    },
    {
        "id": "Q3",
        "ask": "What does p95 hide that the mean does not, and what do 1,000 records cost?",
        "rejected": "It is pretty fast and cheap enough.",
        "hired": (
            f"p50 is {METRICS['p50_ms']} ms; p95 is {METRICS['p95_ms']} ms, so a tail of slow calls "
            f"hides behind the mean. Cost per 1,000 records is ${METRICS['cost_per_1000']:.2f} "
            f"from logged call counts."
        ),
    },
    {
        "id": "Q4",
        "ask": "Name two production failure modes this system hits first, and what you would do.",
        "rejected": "LLMs sometimes hallucinate, so we would monitor it.",
        "hired": (
            "First: out-of-schema labels ('Refunds') — validate at the boundary and fail the row. "
            "Second: silent fallback to General on parse errors — log parse_error, never coerce."
        ),
    },
]

blocks = []
for q in QUESTIONS:
    blocks.append(
        f"### {q['id']}\n\n"
        f"**Question.** {q['ask']}\n\n"
        f"**Rejected.** {q['rejected']}\n\n"
        f"**Hired.** {q['hired']}"
    )
md("\n\n".join(blocks))
```
