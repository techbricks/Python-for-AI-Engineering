## Publish checklist

You are not implementing CI or containers here. Those are next-step options. Today: the remote exists, the tag exists, `.env` is not on the remote, and the README clone command is the same command you used in the audit.


```python
from IPython.display import display, Markdown

def md(text):
    """Display text as rendered markdown."""
    display(Markdown(text))
```

```python
PUBLISH = [
    ("remote", "git remote -v shows origin", True),
    ("tag", "tag capstone or v5 exists on the commit that passed the audit", True),
    ("ignore", ".env is gitignored and git ls-files .env is empty", True),
    ("readme", "README clone, venv, pytest, and CLI commands match what you actually ran", True),
    ("limitations", "LIMITATIONS.md is on that commit", True),
    ("audit_log", "RELEASE_AUDIT.md overall PASS is on that commit", True),
]

lines = ["| Check | Meaning | Ready? |", "|---|---|---|"]
for key, meaning, ok in PUBLISH:
    lines.append(f"| {key} | {meaning} | {'yes' if ok else 'no'} |")
md("### Publish board\n\n" + "\n".join(lines) +
   "\n\nIf any row is no, fix it before you send the link. Pushing is `git push origin HEAD` and `git push origin --tags`.")
```

## Eight next steps — pick exactly three, with dates

These are options, not homework for this notebook.


```python
OPTIONS = [
    "Grow the gold set with newly labelled tickets",
    "Add CI on push",
    "Containerise the CLI",
    "Ship the optional HTTP endpoint",
    "Swap in a local backend for privacy-sensitive tickets",
    "Add cost and latency alerts from structured logs",
    "Human review for low-confidence labels",
    "Apply the same pipeline to a second domain (bug reports)",
]

# Representative three — replace with yours
PICKS = [
    {"option": OPTIONS[0], "date": "2026-10-06", "why": "Account vs General is still the open Module 1 question."},
    {"option": OPTIONS[6], "date": "2026-10-13", "why": "Low-confidence routing was question 4; still open."},
    {"option": OPTIONS[4], "date": "2026-10-20", "why": "Hosted calls plus ticket text is the privacy leftover."},
]

assert len(PICKS) == 3
assert len({p["option"] for p in PICKS}) == 3
assert all(p["option"] in OPTIONS for p in PICKS)

all_lines = ["| # | Option | Picked? |", "|---|---|---|"]
for i, opt in enumerate(OPTIONS, 1):
    mark = "yes" if opt in {p["option"] for p in PICKS} else ""
    all_lines.append(f"| {i} | {opt} | {mark} |")

pick_lines = ["| When | What | Why |", "|---|---|---|"]
for p in PICKS:
    pick_lines.append(f"| {p['date']} | {p['option']} | {p['why']} |")

md(
    "### Menu\n\n" + "\n".join(all_lines)
    + "\n\n### Dated picks\n\n" + "\n".join(pick_lines)
)
```
