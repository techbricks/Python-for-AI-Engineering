## Why this order

A hiring manager spends seconds. They ask four questions in a fixed sequence:

1. What does this do?
2. Can I run it?
3. Did it work?
4. What are the catches?

If the README answers those in any other order — theory first, setup on page four, no numbers — they leave. Alt text is part of the same contract: a reader who cannot see the diagram still gets the relationships.


````python
from IPython.display import display, Markdown

def md(text):
    """Display text as rendered markdown."""
    display(Markdown(text))

from pathlib import Path
````

## Part 1: Score a README that buries the run

> 🧩 **README structure.** Answer what it does, can I run it, did it work, and the catches — in that order.
> Watch how the bad README answers those four questions in the wrong order.

> 🧩 **responsible AI limitations.** The written limitations statement from the video is what the catches question checks for.

This is a plausible bad README for **this** course's product, not a generic RAG toy. Score it against the four-question order.

````python
# A short theory-first README: it talks about history before saying what the repo does
BAD_README = """
# Support Triage

## Background

Customer support has existed for decades. Large language models can classify text.
This repository explores that idea, with a literature review and notes on future work.

## Installation

See source.
"""

# The four reader questions, in the order a hiring manager asks them
QUESTIONS = [
    ("what", "What does this do? (one sentence, who it is for)"),
    ("run", "Can I run it? (clone, install, env, one command)"),
    ("worked", "Did it work? (gold-set number, not 'seems good')"),
    ("catches", "Catches? (scope, failure, cost — a pointer is enough here)"),
]

def score_readme(text: str) -> dict[str, bool]:
    """Crude keyword check: does the README answer each reader question?"""
    lower = text.lower()
    return {
        "what": "triage" in lower and "support" in lower and "background" not in lower[:80],
        "run": "clone" in lower or "pip install" in lower or "uv " in lower,
        "worked": "gold" in lower or "precision" in lower or "recall" in lower or "p95" in lower,
        "catches": "limitation" in lower or "demonstration" in lower or "not a production" in lower,
    }

bad_scores = score_readme(BAD_README)
lines = []
for key, question in QUESTIONS:
    mark = "yes" if bad_scores[key] else "no"
    lines.append(f"- {question}: **{mark}**")

md("### Scoring the theory-first README\n\n" + "\n".join(lines))
````

## Part 2: Fill the template with real numbers

Results must cite a gold set (and its size caveat) and end with a ship sentence. Catches here are one paragraph; the full LIMITATIONS file is the next lab.

Replace the metrics if you have your own. The structure stays.


````python
# Replace these with your own numbers if you have them
GOLD_N = 50
PRECISION_BILLING = 0.84
RECALL_BILLING = 0.71
BASELINE_ACC = 0.62
P95_MS = 2100
COST_PER_1000 = 0.42

# The ship sentence names the win and the accepted miss
SHIP_SENTENCE = (
    f"I would ship this as a demonstration router for Billing vs other because "
    f"Billing precision is {PRECISION_BILLING:.0%} on a {GOLD_N}-row gold set "
    f"(baseline accuracy {BASELINE_ACC:.0%}), accepting that p95 is {P95_MS} ms "
    f"and General is still confused with Account."
)

# One command per line, so a stranger can copy and paste
run_block = "\n".join([
    "git clone https://github.com/YOU/support-triage.git",
    "cd support-triage",
    "pip install -r requirements.txt",
    "cp .env.example .env   # Windows: copy .env.example .env",
    "python -m triage.cli data/support_inbox.csv",
])

# Baseline first, so the model numbers have something to beat
metrics = "\n".join([
    "| Metric | Value |",
    "|---|---|",
    f"| Majority baseline accuracy | {BASELINE_ACC:.0%} |",
    f"| Billing precision | {PRECISION_BILLING:.0%} |",
    f"| Billing recall | {RECALL_BILLING:.0%} |",
    f"| p95 latency | {P95_MS} ms |",
    f"| Cost per 1,000 records | ${COST_PER_1000:.2f} |",
])

# Sections follow the four reader questions, in order
GOOD_README = "\n".join([
    "# Support-message triage",
    "",
    "## What it does",
    "",
    "Labels incoming support tickets as Billing, Technical, Account, or General with a six-box pipeline; the model is one box, not the product.",
    "",
    "## Can I run it?",
    "",
    "```bash",
    run_block,
    "```",
    "",
    "## Did it work?",
    "",
    f"Gold set: **{GOLD_N} labelled tickets** (small sample - treat percentages as directional).",
    "",
    metrics,
    "",
    SHIP_SENTENCE,
    "",
    "## Catches",
    "",
    "This is a demonstration project, not a production system. English tickets only. See LIMITATIONS.md for failure modes, data rights, and secret handling.",
    "",
    "## Pipeline",
    "",
    "![Six-box triage pipeline with real module filenames](docs/pipeline.png)",
    "",
])

checks = score_readme(GOOD_README)
lines = []
for key, question in QUESTIONS:
    mark = "yes" if checks[key] else "no"
    lines.append(f"- {question}: **{mark}**")

md("### Scoring the filled template\n\n" + "\n".join(lines))
````

## Part 3: Relabel the six-box diagram with real filenames

Module 1 used generic box names. A portfolio diagram names the files a stranger will open.


````python
# Each box names the file a stranger will open
STAGES = [
    ("Ingest", "src/triage/ingest.py", "load CSV"),
    ("Prepare", "src/triage/prepare.py", "clean rows, count drops"),
    ("Call model", "src/triage/clients/", "LLM adapter"),
    ("Validate", "src/triage/validate.py", "schema check"),
    ("Log", "src/triage/logging.py", "structured logs"),
    ("Evaluate", "eval/metrics.md", "metrics report"),
]

lines = [
    "| Stage | File | Relationship |",
    "|---|---|---|",
]
for stage, path, relation in STAGES:
    lines.append(f"| {stage} | `{path}` | {relation} |")

md("### Relabelled six-box map\n\n" + "\n".join(lines))
````

## Part 4: Alt text describes content and relationships

## Part 5: Write the README artifact

The notebook writes `artifacts/README.md` next to this lab so you can copy it into the capstone repo. Always use utf-8 so Windows does not break emoji or dashes.


````python
out_dir = Path("artifacts")
out_dir.mkdir(parents=True, exist_ok=True)
readme_path = out_dir / "README.md"
# Inject the good alt text into the image line for the file on disk
text = GOOD_README.replace(
    "![Six-box triage pipeline with real module filenames](docs/pipeline.png)",
    f"![{GOOD_ALT}](docs/pipeline.png)",
)
readme_path.write_text(text, encoding="utf-8")
md(f"Wrote `{readme_path.as_posix()}` ({readme_path.stat().st_size} bytes, utf-8).")
````

## Limitations and responsible AI

What it gets wrong, data/PII, and licensing.


## Why write it down

A reviewer who finds the failure modes themselves will not hire you. A reviewer who reads LIMITATIONS.md and still understands the numbers will. The required sentence is not humble branding: **this is a demonstration project, not a production system.**


## Part 1: Failure modes for *this* pipeline

Reuse the Module 9 list. Attach each mode to a support-ticket behaviour, not to a generic chatbot.


````python
# The three failure modes most likely to bite this pipeline
FAILURE_MODES = [
    {
        "mode": "Out-of-schema labels",
        "here": "Model returns 'Refunds' instead of one of the four labels",
        "mitigation": "Validate at the boundary; never silently coerce",
    },
    {
        "mode": "Silent fallback",
        "here": "Parser fails and code writes General with no log flag",
        "mitigation": "Log parse_error=true; do not swallow the exception",
    },
    {
        "mode": "Rate limits",
        "here": "Hosted backend 429s on a 1,000-row batch",
        "mitigation": "Capped retry; non-zero CLI exit code when the cap is hit",
    },
]

lines = ["| Failure mode | How it shows up in triage | Mitigation |", "|---|---|---|"]
for row in FAILURE_MODES:
    lines.append(f"| {row['mode']} | {row['here']} | {row['mitigation']} |")
md("### Failure modes mapped to this repo\n\n" + "\n".join(lines))
````

A limitation is usable only if a reader can decide whether to trust this output for their ticket. Every sentence in `LIMITATIONS.md` below is written to pass that test.

## Part 2: Privacy, licensing, and secret hygiene checklist

> 🧩 **Data privacy and licensing.** Handle PII with consent and retention limits, and respect dataset licenses.
> Watch which checklist items are release blockers.

Tick these against the triage repo. A no on secrets is a release blocker. A no on licence means you cannot publish the CSV.

````python
# Each row: (key, rule, status). Set the status by reviewing your own repo.
CHECKLIST = [
    ("pii", "Tickets in the published CSV have no names, emails, phones, or account IDs", "yes"),
    ("consent", "If using your own messages, you wrote them or have permission to publish", "yes"),
    ("retention", "Run logs are local; README says not to upload raw tickets to third-party tools", "yes"),
    ("licence", "Dataset licence allows this use and does not require you to ship customer PII", "yes"),
    ("gitignore", ".env, .venv, and key files are listed in .gitignore", "yes"),
    ("example_env", ".env.example has placeholder names, not real keys", "yes"),
    # Not checked automatically here; the secret scan is built in lab-release-audit
    ("no_hardcoded_key", "Source files have no sk- live keys and no password literals", "manual review (verify in lab-release-audit)"),
]

lines = ["| Check | Rule | Pass? |", "|---|---|---|"]
for key, rule, status in CHECKLIST:
    lines.append(f"| {key} | {rule} | {status} |")

md("### Responsible AI checklist\n\n" + "\n".join(lines))
````

## Part 3: Write LIMITATIONS.md

Utf-8 on Windows. This file is the Catches section, expanded so the README can stay short.

````python
LIMITATIONS = """# Limitations

This is a demonstration project, not a production system.

## Where it is wrong

- **Labels.** Allowed classes are Billing, Technical, Account, and General. Out-of-schema strings are a failure, not a new class.
- **Account vs General.** Account is the weak class and is often labelled General.
- **Language.** English tickets only.
- **Latency and sample size.** p95 was 2100 ms in the logged run, and fifty gold rows are directional. Do not quote these numbers as production performance.

## Data rights and privacy

- Publish only tickets with no PII (names, emails, phones, account IDs).
- Do not commit raw customer exports.
- Dataset licence: course sample CSV, for teaching use. Do not redistribute as a commercial corpus.
- Logs keep a run id and label; user text is redacted.

## Secrets

- API keys live in `.env`, which is gitignored.
- `.env.example` has placeholders only.
- If a key was ever committed, revoke it, rotate it, and rewrite history as taught in Module 4.
"""

out_dir = Path("artifacts")
out_dir.mkdir(parents=True, exist_ok=True)
path = out_dir / "LIMITATIONS.md"
path.write_text(LIMITATIONS, encoding="utf-8")
md(f"Wrote `{path.as_posix()}` ({path.stat().st_size} bytes). The required demonstration sentence is line 3.")
````
