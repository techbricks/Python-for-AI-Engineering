| Baseline Comparison | Judge output against a simple reference baseline (a cheap, clear alternative), not against "looks good". If you cannot beat the baseline, you have not earned complexity. |


## Same 9 rows, new evidence column

You marked these rows at the start of the module. Score them again. Every non-zero cell needs a pointer: a path, a metric, or a command output.

Optional peer swap: if you have a partner, spend three minutes on their table and three on yours. Solo is fine — skip the swap.


```python
from IPython.display import display, Markdown

def md(text):
    """Display text as rendered markdown."""
    display(Markdown(text))
```

```python
BASELINE = [
    {"id": 1, "name": "Pipeline runs end to end", "baseline": 2},
    {"id": 2, "name": "Reusable package", "baseline": 2},
    {"id": 3, "name": "Messy-data ingestion", "baseline": 2},
    {"id": 4, "name": "Reproducible repo", "baseline": 1},
    {"id": 5, "name": "Replaceable AI interface", "baseline": 2},
    {"id": 6, "name": "Hardened runtime", "baseline": 1},
    {"id": 7, "name": "Tests that can fail", "baseline": 1},
    {"id": 8, "name": "Evaluation with numbers", "baseline": 2},
    {"id": 9, "name": "Delivered and honest", "baseline": 0},
]

# After README, LIMITATIONS, and the release audit — still two weak engineering rows
NOW = [
    {"id": 1, "score": 2, "evidence": "src/triage/pipeline.py runs labels on data/sample.csv"},
    {"id": 2, "score": 2, "evidence": "from triage.pipeline import run works from tests/"},
    {"id": 3, "score": 2, "evidence": "prepare.py logs dropped empty=4, too_long=2"},
    {"id": 4, "score": 2, "evidence": "RELEASE_AUDIT.md: documented_commands PASS, pytest PASS"},
    {"id": 5, "score": 2, "evidence": "clients/base.py Protocol; fake adapter in tests"},
    {"id": 6, "score": 1, "evidence": "validate.py exists; bounded retry still missing a wall-clock cap"},
    {"id": 7, "score": 1, "evidence": "test_parser.py passes; no timeout fake-client test yet"},
    {"id": 8, "score": 2, "evidence": "50-row gold set; Billing precision 84% vs baseline 62%"},
    {"id": 9, "score": 2, "evidence": "README four-question order; LIMITATIONS.md demonstration sentence"},
]

assert [r["id"] for r in BASELINE] == [r["id"] for r in NOW]

lines = [
    "| # | Criterion | Baseline | Now | Delta | Evidence |",
    "|---|---|---|---|---|---|",
]
weak = []
for old, new in zip(BASELINE, NOW):
    delta = new["score"] - old["baseline"]
    delta_s = f"+{delta}" if delta > 0 else str(delta)
    lines.append(
        f"| {old['id']} | {old['name']} | {old['baseline']} | {new['score']} | {delta_s} | {new['evidence']} |"
    )
    weak.append((new["score"], old["id"], old["name"], new["evidence"]))

weak_sorted = sorted(weak, key=lambda row: (row[0], row[1]))
weakest_two = weak_sorted[:2]
baseline_total = sum(r["baseline"] for r in BASELINE)
now_total = sum(r["score"] for r in NOW)

md(
    "### Rubric self-review\n\n"
    + "\n".join(lines)
    + f"\n\n**Baseline {baseline_total} / 18 → now {now_total} / 18.**"
)
```

## Fix the two weakest rows

Do not start a new sprint. Each fix is a small, evidenced change. After you make it in the real repo, bump the score.


```python
fixes = []
for score, rid, name, evidence in weakest_two:
    if rid == 6:
        action = "Add a max_elapsed_s argument to retry() and a test that fails when the cap is removed."
    elif rid == 7:
        action = "Add FakeClient.timeout() and assert the pipeline raises; confirm the test fails if you delete the raise."
    else:
        action = "Write the missing evidence pointer, then the smallest code or doc change that makes it true."
    fixes.append({"id": rid, "name": name, "score_now": score, "action": action})

lines = ["| Weak row | Current score | Concrete fix |", "|---|---|---|"]
for fix in fixes:
    lines.append(f"| {fix['id']} {fix['name']} | {fix['score_now']} | {fix['action']} |")

md(
    "### Two fixes\n\n"
    + "\n".join(lines)
    + "\n\nA row that cannot name a file is still a 0, even if you remember doing the work."
)
```
