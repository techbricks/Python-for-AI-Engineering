## Why this order

A hiring manager spends seconds. They ask four questions in a fixed sequence:

1. What does this do?
2. Can I run it?
3. Did it work?
4. What are the catches?

If the README answers those in any other order — theory first, setup on page four, no numbers — they leave. Alt text is part of the same contract: a reader who cannot see the diagram still gets the relationships.


````python
from IPython.display import display, Markdown

def md(text):
    """Display text as rendered markdown."""
    display(Markdown(text))

from pathlib import Path
````

## Part 1: Score a README that buries the run

> 🧩 **README structure.** Answer what it does, can I run it, did it work, and the catches — in that order.
> Watch how the bad README answers those four questions in the wrong order.

This is a plausible bad README for **this** course's product, not a generic RAG toy. Score it against the four-question order.


````python
BAD_README = """
# Support Triage

## Background

Customer support has existed for decades. Large language models, first popularised
after the transformer paper, can classify text. This repository explores that idea
in depth, including the history of intent detection, a literature review, and notes
on future work.

## Architecture musings

We considered several abstract taxonomies for pipelines before settling on a design.

## Installation

See source.

## License
MIT
"""

QUESTIONS = [
    ("what", "What does this do? (one sentence, who it is for)"),
    ("run", "Can I run it? (clone, install, env, one command)"),
    ("worked", "Did it work? (gold-set number, not 'seems good')"),
    ("catches", "Catches? (scope, failure, cost — a pointer is enough here)"),
]

def score_readme(text: str) -> dict[str, bool]:
    lower = text.lower()
    return {
        "what": "triage" in lower and "support" in lower and "background" not in lower[:80],
        "run": "clone" in lower or "pip install" in lower or "uv " in lower,
        "worked": "gold" in lower or "precision" in lower or "recall" in lower or "p95" in lower,
        "catches": "limitation" in lower or "demonstration" in lower or "not a production" in lower,
    }

bad_scores = score_readme(BAD_README)
lines = ["| Reader question | Present in the bad README? |", "|---|---|"]
for key, question in QUESTIONS:
    mark = "yes" if bad_scores[key] else "no"
    lines.append(f"| {question} | {mark} |")

md("### Scoring the theory-first README\n\n" + "\n".join(lines) +
   "\n\nFour no's. That README does not earn a clone.")
````

## Part 2: Fill the template with real numbers

Results must cite a gold set (and its size caveat) and end with a ship sentence. Catches here are one paragraph; the full LIMITATIONS file is the next lab.

Replace the metrics if you have your own. The structure stays.


````python
GOLD_N = 50
PRECISION_BILLING = 0.84
RECALL_BILLING = 0.71
BASELINE_ACC = 0.62
P50_MS = 420
P95_MS = 2100
COST_PER_1000 = 0.42

SHIP_SENTENCE = (
    f"I would ship this as a demonstration router for Billing vs other because "
    f"Billing precision is {PRECISION_BILLING:.0%} on a {GOLD_N}-row gold set "
    f"(baseline accuracy {BASELINE_ACC:.0%}), accepting that p95 is {P95_MS} ms "
    f"and General is still confused with Account."
)

run_block = "\n".join([
    "git clone https://github.com/YOU/support-triage.git",
    "cd support-triage",
    "python -m venv .venv",
    "# Windows: .venv\\Scripts\\activate",
    "# macOS or Linux: source .venv/bin/activate",
    "pip install -r requirements.txt",
    "copy .env.example .env   # Windows",
    "# cp .env.example .env   # macOS or Linux",
    "python -m triage.cli data/support_inbox.csv",
])
metrics = "\n".join([
    "| Metric | Value |",
    "|---|---|",
    f"| Majority baseline accuracy | {BASELINE_ACC:.0%} |",
    f"| Billing precision | {PRECISION_BILLING:.0%} |",
    f"| Billing recall | {RECALL_BILLING:.0%} |",
    f"| p50 latency | {P50_MS} ms |",
    f"| p95 latency | {P95_MS} ms |",
    f"| Cost per 1,000 records | ${COST_PER_1000:.2f} |",
])
GOOD_README = "\n".join([
    "# Support-message triage",
    "",
    "Label incoming support tickets as Billing, Technical, Account, or General.",
    "",
    "## What it does",
    "",
    "A six-box pipeline reads a CSV of support messages, prepares messy rows with counted drops,",
    "calls a model behind an interface, validates the label, logs the run, and writes results.",
    "The model is one box, not the product.",
    "",
    "## Can I run it?",
    "",
    "```bash",
    run_block,
    "```",
    "",
    "You need a Python 3.11 interpreter and, for the hosted backend, an API key in `.env` (never committed).",
    "",
    "## Did it work?",
    "",
    f"Gold set: **{GOLD_N} labelled tickets** (small sample - treat percentages as directional).",
    "",
    metrics,
    "",
    SHIP_SENTENCE,
    "",
    "## Catches",
    "",
    f"This is a demonstration project, not a production system. English tickets only. The gold set is {GOLD_N} rows. See LIMITATIONS.md for failure modes, data rights, and secret handling.",
    "",
    "## Pipeline",
    "",
    "![Six-box triage pipeline with real module filenames](docs/pipeline.png)",
    "",
])

checks = score_readme(GOOD_README)
lines = ["| Reader question | Present in the filled README? |", "|---|---|"]
for key, question in QUESTIONS:
    mark = "yes" if checks[key] else "no"
    lines.append(f"| {question} | {mark} |")

md("### Scoring the filled template\n\n" + "\n".join(lines) +
   "\n\nAll four reader questions are answered in order.")
````

## Part 3: Relabel the six-box diagram with real filenames

Module 1 used generic box names. A portfolio diagram names the files a stranger will open.


````python
STAGES = [
    ("Ingest", "src/triage/ingest.py", "Load the CSV with pathlib and utf-8"),
    ("Prepare", "src/triage/prepare.py", "Drop empty and too-long rows; print counts"),
    ("Call model", "src/triage/clients/", "Protocol plus hosted, local, or fake adapter"),
    ("Validate", "src/triage/validate.py", "Check label is in the allowed set"),
    ("Log", "src/triage/logging.py", "Run id, per-row fields, redacted text"),
    ("Evaluate", "eval/metrics.md", "Gold set, precision, recall, p95, cost"),
]

lines = [
    "| Stage | File | Relationship |",
    "|---|---|---|",
]
for stage, path, relation in STAGES:
    lines.append(f"| {stage} | `{path}` | {relation} |")

md("### Relabelled six-box map\n\n" + "\n".join(lines))
````

## Part 4: Alt text describes content and relationships

> 🧩 **Alt text for your diagram.** Explain the content and relationships the diagram conveys, not how it looks.
> Watch how the bad alt text narrates appearance instead of the claim.

Bad alt text narrates paint. Good alt text narrates the claim the diagram makes.


````python
BAD_ALT = "A screenshot of a blue architecture diagram with rounded boxes."
GOOD_ALT = (
    "Diagram of the support-triage pipeline: ingest.py loads a CSV, prepare.py drops "
    "invalid rows with counts, a Protocol client calls the model, validate.py checks "
    "the label, logging.py records a redacted run, and eval/metrics.md reports gold-set "
    "precision, p95 latency, and cost per 1,000 records."
)

md(
    "### Alt text comparison\n\n"
    f"**Bad (appearance):** {BAD_ALT}\n\n"
    f"**Good (content and relationships):** {GOOD_ALT}\n\n"
    "The good version never mentions colour, shape, or 'screenshot'. It says what "
    "depends on what."
)
````

## Part 5: Write the README artifact

The notebook writes `artifacts/README.md` next to this lab so you can copy it into the capstone repo. Always use utf-8 so Windows does not break emoji or dashes.


````python
out_dir = Path("artifacts")
out_dir.mkdir(parents=True, exist_ok=True)
readme_path = out_dir / "README.md"
# Inject the good alt text into the image line for the file on disk
text = GOOD_README.replace(
    "![Six-box triage pipeline with real module filenames](docs/pipeline.png)",
    f"![{GOOD_ALT}](docs/pipeline.png)",
)
readme_path.write_text(text, encoding="utf-8")
md(f"Wrote `{readme_path.as_posix()}` ({readme_path.stat().st_size} bytes, utf-8).")
````
