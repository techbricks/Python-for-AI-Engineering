## The Capstone Brief

You are shipping **one product**: a support-message triage pipeline. It reads messy tickets, labels each one (Billing, Technical, Account, General), and exposes that core through a command-line interface.

This is a **demonstration project**, not a production router. The hire-ready version has tests, numbers, and honest limitations. The **minimum viable version is enough to pass**: the six boxes run, a stranger can clone and run the documented commands, and the README plus LIMITATIONS are honest.

You will not start a new product. You will finish the triage repo you have been building since Module 1.


```python
from IPython.display import display, Markdown

def md(text):
    """Display text as rendered markdown."""
    display(Markdown(text))
```

## Step 1: Choose Dataset and Scope

Two dataset options, both non-confidential:

1. **Course CSV** — the support-inbox rows you have used since Module 1. Default. No PII by design.
2. **Your own messages** — only if they contain no names, emails, phone numbers, or account IDs.

Scope is which pipeline boxes you will **finish** versus **leave as stretch**. Minimum viable version: ingest through deliver run on real rows, plus a clone-and-run story. Stretch: counted cleaning, replaceable interface, tests, gold-set numbers, CLI polish.


```python
DATASET_CHOICE = "course_csv"  # or "own_non_pii_messages"

SCOPE = {
    "ingest": "mvv",
    "prepare": "mvv",
    "call_model": "mvv",
    "validate": "stretch",
    "log": "stretch",
    "evaluate": "mvv",
}

assert DATASET_CHOICE in {"course_csv", "own_non_pii_messages"}
assert set(SCOPE) == {"ingest", "prepare", "call_model", "validate", "log", "evaluate"}

rows = ["| Stage | Scope |", "|---|---|"]
for stage, level in SCOPE.items():
    label = "Minimum viable" if level == "mvv" else "Stretch"
    rows.append(f"| {stage.replace('_', ' ')} | {label} |")

md(
    f"**Dataset:** `{DATASET_CHOICE}`\n\n"
    "A baseline comparison needs a written scope. If a box is stretch, a missing test there "
    "does not fail the module — it is a gap you will score honestly on the rubric.\n\n"
    + "\n".join(rows)
)
```

## Step 2: Mark the Rubric Baseline

Score each row `0` (missing), `1` (partial), or `2` (evidenced in the repo). This is a **baseline**, the same idea as comparing model output to a dumb reference: you are comparing *this repo, today* to the hire-ready bar, not to a vibe.

You already built most of this in Modules 1 through 10. The ship-it module adds documentation, an audit, and a spoken trade-off. Row 9 is usually the weakest on day one of this module.


```python
RUBRIC = [
    {"id": 1, "name": "Pipeline runs end to end", "evidence_hint": "Six boxes produce labelled output (C0)", "score": 2},
    {"id": 2, "name": "Reusable package", "evidence_hint": "Importable module, no hidden notebook state (C1)", "score": 2},
    {"id": 3, "name": "Messy-data ingestion", "evidence_hint": "Counted drops on real rows (C2)", "score": 2},
    {"id": 4, "name": "Reproducible repo", "evidence_hint": "Fresh clone, pins, gitignore (C3)", "score": 1},
    {"id": 5, "name": "Replaceable AI interface", "evidence_hint": "Protocol seam; mock, hosted, or local swap (C4)", "score": 2},
    {"id": 6, "name": "Hardened runtime", "evidence_hint": "Boundary validation, bounded retry, structured logs (C5)", "score": 1},
    {"id": 7, "name": "Tests that can fail", "evidence_hint": "Unit tests plus a fake client for bad JSON or timeout (C6)", "score": 1},
    {"id": 8, "name": "Evaluation with numbers", "evidence_hint": "Gold set, precision and recall vs baseline, p95, cost per 1,000 (C7)", "score": 2},
    {"id": 9, "name": "Delivered and honest", "evidence_hint": "CLI contract, README order, LIMITATIONS, no secrets (C8-C9)", "score": 0},
]

for row in RUBRIC:
    assert row["score"] in {0, 1, 2}

baseline_total = sum(row["score"] for row in RUBRIC)
baseline_max = 2 * len(RUBRIC)

lines = [
    "| # | Criterion | Hint | Score |",
    "|---|---|---|---|",
]
for row in RUBRIC:
    lines.append(
        f"| {row['id']} | {row['name']} | {row['evidence_hint']} | {row['score']} |"
    )

md(
    "### Rubric baseline (before ship-it work)\n\n"
    + "\n".join(lines)
    + f"\n\n**Baseline total: {baseline_total} / {baseline_max}**\n\n"
    "Save this number. The self-review mini-lab compares against it."
)
```
