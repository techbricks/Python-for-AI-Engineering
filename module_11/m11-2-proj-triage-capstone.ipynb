## Concepts Applied

| Concept | Gloss | From |
|---------|-------|------|
| AI Pipeline Stages | A practical AI feature pipeline with six boxes: ingest, prepare, call, validate, log, evaluate. The model call is just one box, not the whole system. | Module 1 |
| Baseline Comparison | Judge output against a simple reference baseline (a cheap, clear alternative), not against "looks good". If you cannot beat the baseline, you have not earned complexity. | Module 1 |
| Counted Cleaning | A data cleaning habit where every step reports how many rows it removed, so you can verify assumptions, spot surprises, and keep an auditable trail of what changed. | Module 3 |
| Reproducibility | Someone else's machine reproduces your result from documented, pinned steps, starting from a fresh repo clone with no hidden setup. | Module 4 |
| Secret Hygiene | Secrets hygiene is the habit of keeping passwords, API keys, and tokens out of code and logs, sharing them safely, and rotating them regularly so one mistake doesn’t become a breach. | Module 4 |
| Interface Seam | A small typed contract between your app and an AI vendor. Your adapter returns your dataclass, never their SDK response object, so providers can be swapped without rewrites. | Module 6 |
| Boundary Validation | Check user input and model output, the parts you do not control, so errors, abuse, and surprises are caught before they hit your core logic. | Module 7 |
| Bounded Retry | Retry transient errors only, and cap attempts, total wall-clock time, and cost or rate-limit budget so failures end fast, predictably, and safely. | Module 7 |
| Structured Logging | Logs as records with a run id and consistent per-record fields, so you can trace behavior and search reliably, while redacting sensitive keys and user text by default. | Module 7 |
| Unit Test | A small assertion about one behavior. A unit test that cannot fail is theatre, it should catch a real bug when the behavior breaks. | Module 8 |
| Test Double | An injected fake backend that can return malformed JSON or time out on demand, so tests reliably exercise error paths without calling real external services. | Module 8 |
| Gold Set | A small, verified labeled sample used as the trusted reference for evaluation. It gives high-signal quality checks, but small sample sizes can mislead, so treat results as estimates, not absolute truth. | Module 9 |
| Precision and Recall | Ways to measure quality by comparing what you flagged vs what really exists. Precision asks, "Of what I flagged, how much was right?" Recall asks, "Of what existed, how much did I catch?" Per-class metrics beat a single headline number. | Module 9 |
| Percentile Latency | The mean hides the slow calls and timeouts, so use percentiles like p50 and p95 to describe typical latency and tail slowness users actually feel. | Module 9 |
| Cost Per Unit Estimate | Compute cost per 1,000 records from your own logged counts, then treat price as a variable so you can compare models and prompts fairly. | Module 9 |
| Command-Line Interface | A CLI contract defines how a command is used and how it behaves: its arguments and flags, what `--help` promises, which exit codes mean success or failure, and what goes to stdout versus stderr. | Module 10 |
